# AI Results from Q learning

```python
from q_learning import QLearningBot
```

```python
ai_results_list: list[AiResults] = []
class AiResults():
    def __init__(self, name, wins, losses, ties):
        self.name = name
        self.wins = wins
        self.losses = losses
        self.ties = ties
```

```python
q_table_list = ['q_tables/q_table_500_training.npy',
'q_tables/q_table_1000_training.npy',
'q_tables/q_table_2500_training.npy',
'q_tables/q_table_5000_training.npy',
'q_tables/q_table_10000_training.npy',
'q_tables/q_table_15000_training.npy',
'q_tables/q_table_25000_training.npy',
'q_tables/q_table_50000_training.npy']

for training_model in q_table_list:

    bot = QLearningBot(300, training_model, False)
    bot.train()
    result = AiResults(training_model, bot.wins, bot.losses, bot.ties)
    ai_results_list.append(result)
```
